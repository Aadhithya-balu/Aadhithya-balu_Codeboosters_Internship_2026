### Volume
- The amount the of the data generated

### Velocity
- The speed of the data retrieval or generated and processed
- Related to flipkart : The platform gets the real-time data and updates the inventory of the user's data and the whole data is processed within seconds

### Variety
- The type or category of data is defined
- Related to flipkart : The data is categorized into several formats like structured, semi-structured and unstructured

### Veracity
- The quality ,trust worthiness and accuracy of the data is mentioned
- Related to flipkart : Flipkart handles users account without any problem , fake customer reviews and corrupt GPS tracking points.

### Value
- The ultimate benefit derived from the analysis of data (Business ready data)
- Related to flipkart : By turning the data into insights , it gets idea of the trend and optimize the platform according to it.

# Medallion Architecture
- It is a data design pattern used to organize data logically 

## Bronze layer
- The data is collected from various sources and stored as a raw file or parquet file
- Raw data lives here

## Silver layer
- The raw data is now transformed into a clean data by performing the functions
- The clean data lives here

## Gold layer
- The business need data is made here by making the useful insights with the cleaned data
- The Insightful or Business ready data lives here


| Feature | Data Lake | Data Warehouse |
| :--- | :--- | :--- |
| **Data Type** | Structured, semi-structured, and unstructured (JSON, logs, audio, video). | Strictly structured (relational tables with predefined schemas). |
| **Schema Design** | **Schema-on-Read**  | **Schema-on-Write** |
| **Processing Speed**| Highly scalable for raw ingestion | Highly optimized for ultra-fast analytical queries and aggregations. |
| **Storage Cost** | Low | High |
| **Key Use Cases** | Big data analytics, Machine Learning, AI training, and raw data archiving. | Business Intelligence (BI), historical reporting, dashboards, and structured SQL analysis. |
